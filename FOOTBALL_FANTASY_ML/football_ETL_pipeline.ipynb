```python
import pandas as pd

seasons = ["2020-2021", "2021-2022", "2022-2023", "2023-2024", "2024-2025", "2025-2026"]

file_types = ["standard", "shooting", "keeper", "playing_time", "misc"]

csvtypes = {
    "standard":     "Big-5-European-Leagues-Stats.csv",
    "shooting":     "Big-5-European-Leagues-Stats-Shooting.csv",
    "keeper":       "Big-5-European-Leagues-Stats-Keeper.csv",
    "playing_time": "Big-5-European-Leagues-Stats-Playing-Time.csv",
    "misc":         "Big-5-European-Leagues-Stats-Misc.csv",
}

dfs = {}
for season in seasons:
    for ftype, suffix in csvtypes.items():
        key = f"{ftype}_{season}"
        dfs[key] = pd.read_csv(f"{season}-{suffix}")
        print(f"[{key}] shape: {dfs[key].shape}")
```

```python
short = "25_26"

all_cols = {
    "standard":     set(globals()[f"standard_{short}"].columns),
    "shooting":     set(globals()[f"shooting_{short}"].columns),
    "keeper":       set(globals()[f"keeper_{short}"].columns),
    "playing_time": set(globals()[f"playing_time_{short}"].columns),
    "misc":         set(globals()[f"misc_{short}"].columns),
}

common = set.intersection(*all_cols.values())
print("Common columns:", common)

for name, cols in all_cols.items():
    print(f"\n{name} only: {sorted(cols - common)}")
```

```python
JOIN_KEYS = ["Player", "Squad", "Born"]

def merge_season(short):
    base = globals()[f"standard_{short}"].copy()

    others = {
        "shooting":     globals()[f"shooting_{short}"],
        "keeper":       globals()[f"keeper_{short}"],
        "playing_time": globals()[f"playing_time_{short}"],
        "misc":         globals()[f"misc_{short}"],
    }

    for ftype, df in others.items():
        base = base.merge(df, on=JOIN_KEYS, how="outer" ,suffixes=("", f"_{ftype}"))

    return base

merged_25_26 = merge_season("25_26")
print(merged_25_26.shape)
merged_25_26.head()
```

```python
merged_25_26.info(verbose=True, show_counts=True)
```

```python
seasons_short = [f"{s.split('-')[0][2:]}_{s.split('-')[1][2:]}" for s in seasons]

merged_dfs = {}
for short in seasons_short:
    merged_dfs[short] = merge_season(short)
    print(f"[{short}] shape: {merged_dfs[short].shape}")
```

```python
for short, season in zip(seasons_short, seasons):
    merged_dfs[short]["season"] = season

final_df = pd.concat(merged_dfs.values(), ignore_index=True)
print(final_df.shape)
final_df.head(20)
```
